```python
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Fill fulfilled-by using Fulfilment column
df['fulfilled-by'] = df['fulfilled-by'].fillna(
    df['Fulfilment'].map({
        'Amazon': 'Blue Dart',
        'Merchant': 'Merchant'
    })
)

# Check the courier distribution
print(df['fulfilled-by'].value_counts())

# Step 2: Total orders by courier
total_orders = (
    df.groupby('fulfilled-by')['Order ID']
      .count()
)

# Step 3: Consider delivered/shipped orders as successful fulfillment
delivered = df[
    df['Status'].str.contains(
        'Delivered|Shipped',
        case=False,
        na=False
    )
]

# Step 4: Successful orders by courier
delivered_orders = (
    delivered.groupby('fulfilled-by')['Order ID']
             .count()
)

# Step 5: Ensure all couriers are included
delivered_orders = delivered_orders.reindex(
    total_orders.index,
    fill_value=0
)

# Step 6: Calculate fulfillment ratio
ratio = (
    delivered_orders / total_orders
) * 100

ratio = ratio.sort_values(ascending=False)

# Display results
print("\nFulfillment Ratio (%)")
print(ratio.round(2))

# Step 7: Visualization
plt.figure(figsize=(8,5))

ax = ratio.plot(kind='bar')

for i, v in enumerate(ratio):
    plt.text(
        i,
        v + 0.5,
        f'{v:.1f}%',
        ha='center'
    )

plt.title('Courier Fulfillment Ratio')
plt.xlabel('Courier')
plt.ylabel('Fulfillment Ratio (%)')
plt.xticks(rotation=0)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Map courier names
df['fulfilled-by'] = df['fulfilled-by'].fillna(
    df['Fulfilment'].map({
        'Amazon': 'Blue Dart',
        'Merchant': 'Merchant'
    })
)

# Cancelled orders
cancelled = df[
    df['Status'].str.contains(
        'Cancelled',
        case=False,
        na=False
    )
]

# Total revenue by courier
total_revenue = (
    df.groupby('fulfilled-by')['Amount']
      .sum()
)

# Cancelled revenue by courier
cancelled_revenue = (
    cancelled.groupby('fulfilled-by')['Amount']
             .sum()
)

# Ensure all couriers appear
cancelled_revenue = cancelled_revenue.reindex(
    total_revenue.index,
    fill_value=0
)

# Loss ratio
loss_ratio = (
    cancelled_revenue / total_revenue
) * 100

# Summary table
loss_summary = pd.DataFrame({
    'Total Revenue': total_revenue,
    'Cancelled Revenue': cancelled_revenue,
    'Loss Ratio (%)': loss_ratio.round(2)
})

print(loss_summary.sort_values('Loss Ratio (%)', ascending=False))

# Visualization
plt.figure(figsize=(8,5))

ax = loss_ratio.sort_values(ascending=False).plot(
    kind='bar'
)

for i, v in enumerate(loss_ratio.sort_values(ascending=False)):
    plt.text(i, v, f'{v:.2f}%', ha='center')

plt.title('Revenue Loss Ratio by Courier')
plt.xlabel('Courier')
plt.ylabel('Loss Ratio (%)')
plt.xticks(rotation=0)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Classify promotion codes
def classify_promo(x):

    x = str(x).upper()

    if 'FREE SHIPPING' in x:
        return 'Free Shipping'

    elif 'PLCC' in x or 'FREE-FINANCING' in x:
        return 'Financing Offer'

    elif 'PRIME' in x:
        return 'Prime Day Campaign'

    elif 'SUMMER' in x:
        return 'Summer Sale'

    elif 'COUPON' in x or 'DISCOUNT' in x:
        return 'Coupon / Discount'

    else:
        return 'Other Promotions'

# Create promotion category column
df['promo_category'] = df['promotion-ids'].apply(classify_promo)

# Promotion usage
promo_usage = (
    df['promo_category']
      .value_counts()
)

print("Promotion Usage:")
print(promo_usage)

# Revenue generated by promotion type
promo_revenue = (
    df.groupby('promo_category')['Amount']
      .sum()
      .sort_values(ascending=False)
)

print("\nRevenue by Promotion Category:")
print(promo_revenue)

# Visualization
plt.figure(figsize=(8,5))

promo_revenue.plot(kind='bar')

plt.title('Revenue Generated by Promotion Category')
plt.xlabel('Promotion Category')
plt.ylabel('Revenue')
plt.xticks(rotation=45)

for i, v in enumerate(promo_revenue):
    plt.text(i, v, f'₹{v:,.0f}', ha='center')

plt.show()
```
